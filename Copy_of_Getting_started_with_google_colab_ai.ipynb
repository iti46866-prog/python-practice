```python
def prime():
  num = int(input("enter a number: "))
  if num == 0 or num ==1:
    print(f"{num} is not a prime number")
  for i in range(2,num):
      if num%i == 0:
       print(f"{num} is not a prime number")
       break
  else:
    print(f"{num} is a prime number")
```

```python
prime()
```

```python
def new_list():
  a =list(map(int,input("enter a list: ").split(",")))
  b=[]
  first_value = a[0]
  last_value = a[-1]
  b.append(first_value)
  b.append(last_value)
  print(b)
```

```python
def fibonacci():
  num = int(input("enter the number of numbers you want to generate: "))
  a =0
  b=1
  fib =[1]
  for i in range(2,num+1):
    c=a+b
    a=b
    b=c
    fib.append(c)

  print(fib)

```

```python
fibonacci()
```

```python
def reverse_order():
  string = str(input("enter a string: "))
  result1=string[::-1]
  print(result1)
  result2 =" ".join(string.split()[::-1])
  print(result2)
```

```python
reverse_order()
```

```python
def password_generator():
  import random
  import string
  characters = string.ascii_letters + string.digits + string.punctuation
  level = str(input("enter the level of password (weak,medium,strong): "))
  if level=="weak":
     length = 5
  elif level =="medium":
    length = 9
  else:
      length = 12

  password = "".join(random.choice(characters) for i in range(length))
  print(password)
```

```python
password_generator()
```

```python
import random
def cow_bull():
  number = str(random.randint(1000,9999))
  guess =1
  while True:
   guess+=1
   num=input("enter a number from (1000,9999): ")
   cow=0
   bull=0
   for i in range(4):
    if num[i]==number[i]:
      cow+=1
    elif num[i] in number:
      bull+=1

   print(f"you got {cow} cows and {bull}bulls")

   if num==number:
    print(f"you got it in {guess} guesses")
    break
```

```python
cow_bull()
```
