```python
def find_num():
  a = list(map(int,input("enter a list: ").split(",")))
  a.sort()
  b = int(input("enter a number: "))
  if b in a:
    print("number is in the list")
  else:
    print("number is not in the list")
```

```python
def findin_list():
  a = list(map(int,input("enter a list: ").split(",")))
  a.sort()
  b = int(input("enter a number: "))
  low =0
  high =len(a)-1
  while low<=high:
    mid = int((low+high)/2)
    if a[mid]==b:
      print("number is in the list")
      return
    elif a[mid]<b:
      low = mid+1
    else:
      high=mid-1

  print("number is not in the list")

```

```python
findin_list()
```

```python
import requests
from bs4 import BeautifulSoup
url = "http://newsrss.bbc.co.uk/rss/newsonline_uk_edition/front_page/rss.xml"
response=requests.get(url)
soup=BeautifulSoup(response.content,"xml")
with open("news.txt","w") as open_file:
 for i, item in enumerate(soup.find_all("item")):
  open_file.write(item.title.text +"\n")
  print(item.title.text)
  open_file.write(item.link.text +"\n")
  print(item.link.text)
```

```python
count={}
with open("nameslist.txt","r") as file:
  for line in file:
    name = line.strip()

    if name in count:
      count[name]+=1
    else:
      count[name]=1

for name,count in count.items():
  print(f"{name}:{count}")
```
