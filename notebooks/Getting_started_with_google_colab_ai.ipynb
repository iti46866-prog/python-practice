```python
def prime():
  num = int(input("enter a number: "))
  if num == 0 or num ==1:
    print(f"{num} is not a prime number")
  for i in range(2,num):
      if num%i == 0:
       print(f"{num} is not a prime number")
       break
  else:
    print(f"{num} is a prime number")
```

```python
prime()
```

```python
def new_list():
  a =list(map(int,input("enter a list: ").split(",")))
  b=[]
  first_value = a[0]
  last_value = a[-1]
  b.append(first_value)
  b.append(last_value)
  print(b)
```

```python
new_list()
```

```python
def fibonacci():
  num = int(input("enter the number of numbers you want to generate: "))
  a =0
  b=1
  fib =[1]
  for i in range(2,num+1):
    c=a+b
    a=b
    b=c
    fib.append(c)

  print(fib)

```

```python
fibonacci()
```

```python
def remove_duplicates():
  a = list(map(int,input("enter a list: ").split(",")))
  b=[]
  b= set(a)
  b=list(b)

  print(b)
```

```python
def reverse_order():
  string = str(input("enter a string: "))
  result1=string[::-1]
  print(result1)
  result2 =" ".join(string.split()[::-1])
  print(result2)
```

```python
reverse_order()
```
